**Ejercicio 5**: Escribir una clase mazo, que construya el mazo de cartas españolas. Escribir un método que devuelve cuantas cartas hay en el mazo.

```python
import random

class Carta:
    apodos = {
        1: "Ancho",
        10: "Sota",
        11: "Caballo",
        12: "Rey"
    }

    def __init__(self, numero, palo):
        self.numero = numero
        self.palo = palo

    def __str__(self):
        if self.numero in Carta.apodos:
            return Carta.apodos[self.numero] + " de " + self.palo
        else:
            return f"{self.numero} de {self.palo}"

    def __eq__(self, other):
        return self.numero == other.numero


class Mazo:
    def __init__(self):
        palos = ["Espada", "Basto", "Copa", "Oro"]
        self.cartas = []

        for n in [1,2,3,4,5,6,7,10,11,12]:
            for p in palos:
                self.cartas.append(Carta(n, p))

    def cantidad(self):
        return len(self.cartas)

    def mezclar(self):
        random.shuffle(self.cartas)
        print("Mezclado")

    def sacar_carta(self):
        carta_sacada = self.cartas.pop()
        return carta_sacada
```

```python
Mazotest = Mazo()

Mazotest.mezclar()

for carta in Mazotest.cartas:
    print(carta)

print("\nCarta sacada:")
print(Mazotest.sacar_carta())

print("\nCantidad restante:")
print(Mazotest.cantidad())
```
